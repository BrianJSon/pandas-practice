  # Answering these analytical questions:
    1.	Which department has the highest average salary?
    2.	What is the average age of employees?
    3.	Which city has the most employees?
    4.	What percentage of employees are active?
    5.	Is there a relationship between employee scores and salaries?

```python
import pandas as pd

df_active = pd.read_csv('employees/data/employees_active.csv')
df_clean = pd.read_csv('employees/data/employees_cleaned.csv')
```

```python
# 1	Which department has the highest average salary?
avg_salary = df_active.groupby('department')['salary'].mean().sort_values(ascending=False)

department = avg_salary.index[0]
salary = avg_salary.iloc[0]
# Marketing has highest average salary of $126500.00
print(f"{department} has the highest average salary of ${salary:,.2f}")
```

```python
# 2	What is the average age of employees?
avg_age = df_active['age'].mean().round(2)
print(f"Average age of employees is {avg_age:.2f} years old.")
```

```python
# 3	Which city has the most employees?
city_counts = df_active['city'].value_counts()

city = city_counts.index[0]
count = city_counts.iloc[0]

print(f"{city} has the most employees with {count} employees")
```

```python
# 4	What percentage of employees are active?
total_employees = len(df_clean)
active_employees = len(df_clean[df_clean['active'] == True])

percentage = (active_employees / total_employees) * 100
print(f"{percentage:.2f}% of employees are active.")
```

```python
# 5	Is there a relationship between employee scores and salaries?
correlation = df_active['score'].corr(df_active['salary'])
print(correlation)
```

```python
# data visualization for question 5
import matplotlib.pyplot as plt

plt.scatter(df_active['score'], df_active['salary'])

plt.xlabel('Employee Score')
plt.ylabel('Employee Salary')
plt.title('Employee Score vs Salary')

plt.show()
print("There is not a strong linear relationship between an employee's score and their salary. Higher scores do not necessarily mean a higher salary, meaning there are other factors contributing to an employee's salary.")
```
