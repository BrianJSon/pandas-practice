```python
import pandas as pd

'''
  Answering these analytical questions:
    1	Which department has the highest average salary?
    2	What is the average age of employees?
    3	Which city has the most employees?
    4	What percentage of employees are active?
    5	Is there a relationship between employee scores and salaries?
'''

file_path = '/content/drive/MyDrive/Colab Notebooks/data/employees.csv'
df = pd.read_csv(file_path)
```

```python
''' formatting the dataset '''

# drop duplicates - email
before = len(df)

df['city'] = df['city'].str.strip().str.title()
df['email'] = df['email'].str.strip().str.lower()

has_email = df[df['email'].notna()]
no_email = df[df['email'].isna()]
has_email = has_email.drop_duplicates(subset='email', keep='first')

df_clean = pd.concat([has_email, no_email])
removed1 = before - len(df_clean)
df_clean
```

```python
# drop duplicates - same name + city + department + age
df_clean = df_clean.drop_duplicates(subset=['name', 'city', 'department', 'age'], keep='first')
df_clean = df_clean.reset_index(drop=True)
removed2 = before - len(df_clean)
df_clean
```

```python
print(f"dropped {removed1} duplicate emails")
print(f"dropped {removed1 - removed2} more")
print(f"dropped {removed2} in total")
```

```python
print(f"Removed {round(len(df_clean) / len(df) * 100, 2)}%")
print(f"Total emails: {len(df['email'])}")
print(f"Unique emails: {len(df['email'].unique())}")
print(f"Missing emails: {df['email'].isna().sum()}")
```

```python
potential_duplicates = df_clean[
    df_clean.duplicated(subset=['name', 'city', 'department'], keep=False)
].sort_values(by=['name', 'city', 'department'])

potential_duplicates
```

```python
df_active = df_clean[df_clean['active'] == True]
df_active
```

```python
df_clean.to_csv('employees_cleaned.csv', index=False)
df_active.to_csv('employees_active.csv', index=False)
```
